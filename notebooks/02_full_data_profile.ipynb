# Full dataset quality profile

Results from **every row** of all four combined CSVs. Run `python pipeline/profile_full.py` from the project root to refresh the reports, then rerun this notebook. No source rows or columns are removed.

2025 is reserved for held-out evaluation; inspect its structural quality and coverage without using it to tune features or models. The role of 2024 remains undecided.

All fields were read as strings to preserve identifiers. Numeric compatibility and ranges are assessed separately. See `reports/full_profile/README.md` for null conventions and limitations.

```python
from pathlib import Path
import json
import pandas as pd
from IPython.display import display

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "AGENTS.md").exists())
RESULTS = ROOT / "reports" / "full_profile"
summary = json.loads((RESULTS / "summary.json").read_text())
pd.set_option("display.max_columns", None)

def show(filename):
    with pd.option_context("display.max_rows", None):
        display(pd.read_csv(RESULTS / filename))

display(pd.DataFrame([
    {"dataset": name, "rows": info["rows"], "columns": info["columns"],
     "exact_duplicate_extra_rows": info["duplicates"]["exact_rows"]["extra_rows"],
     "repeated_key_extra_rows": info["duplicates"]["candidate_keys"]["extra_rows"],
     "rows_missing_keys": info["rows_missing_key_fields"]}
    for name, info in summary.items()
]))
```

### Missingness by event type (%)

```python
events = pd.read_csv(RESULTS / "storm_details_by_event_type.csv")
missing_cols = [c for c in events if c.endswith("__null")]
percentages = events[missing_cols].div(events["rows"], axis=0).mul(100).round(2)
percentages.columns = percentages.columns.str.removesuffix("__null")
with pd.option_context("display.max_rows", None):
    display(pd.concat([events[["EVENT_TYPE", "rows"]], percentages], axis=1))
```
